```python
import pandas as pd


diabetes_url = 'https://raw.githubusercontent.com/plotly/datasets/master/diabetes.csv'
!wget -O diabetes.csv {diabetes_url}


df_diabetes = pd.read_csv('diabetes.csv')


print("--- Diabetes Dataset Head ---")
display(df_diabetes.head())

print("\n--- Column Information ---")
df_diabetes.info()
```

```python
import numpy as np


columns_with_zeros = ['Glucose', 'BloodPressure', 'SkinThickness', 'Insulin', 'BMI']
print("--- Counts of zeros in relevant columns ---")
print((df_diabetes[columns_with_zeros] == 0).sum())


df_diabetes[columns_with_zeros] = df_diabetes[columns_with_zeros].replace(0, np.nan)


df_diabetes[columns_with_zeros] = df_diabetes[columns_with_zeros].fillna(df_diabetes[columns_with_zeros].median())

print("\n--- Missing values after handling ---")
print(df_diabetes.isnull().sum())

categorical_cols = df_diabetes.select_dtypes(include=['object', 'category']).columns.tolist()
print(f"\nCategorical columns identified: {categorical_cols}")


def handle_outliers_iqr(df, columns):
    for col in columns:
        Q1 = df[col].quantile(0.25)
        Q3 = df[col].quantile(0.75)
        IQR = Q3 - Q1
        lower_bound = Q1 - 1.5 * IQR
        upper_bound = Q3 + 1.5 * IQR

        df[col] = np.clip(df[col], lower_bound, upper_bound)
    return df
numerical_cols = df_diabetes.columns.drop('Outcome')
df_diabetes_cleaned = handle_outliers_iqr(df_diabetes.copy(), numerical_cols)
print("\nOutliers handled using IQR clipping.")
```

```python
from sklearn.preprocessing import MinMaxScaler, StandardScaler

features = df_diabetes_cleaned.columns.drop('Outcome')


min_max_scaler = MinMaxScaler()
df_minmax = df_diabetes_cleaned.copy()
df_minmax[features] = min_max_scaler.fit_transform(df_diabetes_cleaned[features])

print("--- Data after Min-Max Scaling (Sample) ---")
display(df_minmax.head())


standard_scaler = StandardScaler()
df_standardized = df_diabetes_cleaned.copy()
df_standardized[features] = standard_scaler.fit_transform(df_diabetes_cleaned[features])

print("\n--- Data after Standardization (Sample) ---")
display(df_standardized.head())

print("\nMin-Max Range (Expected 0 to 1):")
print(df_minmax[features].agg(['min', 'max']))

print("\nStandardization Summary (Expected Mean ~0, Std ~1):")
print(df_standardized[features].agg(['mean', 'std']).round(2))
```

```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import LabelEncoder, MinMaxScaler, StandardScaler
url = 'https://archive.ics.uci.edu/ml/machine-learning-databases/adult/adult.data'
column_names = ['age', 'workclass', 'fnlwgt', 'education', 'education-num', 'marital-status',
                'occupation', 'relationship', 'race', 'sex', 'capital-gain', 'capital-loss',
                'hours-per-week', 'native-country', 'income']


df_adult = pd.read_csv(url, names=column_names, skipinitialspace=True, na_values='?')


print("--- Columns with missing values (count > 0) ---")
missing_counts = df_adult.isnull().sum()
print(missing_counts[missing_counts > 0])


for col in df_adult.columns:
    if df_adult[col].isnull().any():
        if df_adult[col].dtype == 'object':
            df_adult[col] = df_adult[col].fillna(df_adult[col].mode()[0])
        else:
            df_adult[col] = df_adult[col].fillna(df_adult[col].median())


le = LabelEncoder()
cat_cols = df_adult.select_dtypes(include=['object']).columns
for col in cat_cols:
    df_adult[col] = le.fit_transform(df_adult[col])

num_cols = df_adult.select_dtypes(include=[np.number]).columns
for col in num_cols:
    Q1 = df_adult[col].quantile(0.25)
    Q3 = df_adult[col].quantile(0.75)
    IQR = Q3 - Q1
    df_adult[col] = np.clip(df_adult[col], Q1 - 1.5 * IQR, Q3 + 1.5 * IQR)


mms = MinMaxScaler()
ss = StandardScaler()

df_adult_minmax = df_adult.copy()
df_adult_minmax[num_cols] = mms.fit_transform(df_adult[num_cols])

df_adult_std = df_adult.copy()
df_adult_std[num_cols] = ss.fit_transform(df_adult[num_cols])

print("\n--- Preprocessing and Scaling Complete ---")
print("Sample of Min-Max Scaled Numerical Data:")
display(df_adult_minmax[num_cols].head())
print("\nSample of Standardized Numerical Data:")
display(df_adult_std[num_cols].head())
```
